```python
#C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true. 
import numpy as np

X = np.array([[1, 2],
              [3, 4],
              [5, 6]])

w = np.array([0.5, 1.0])
b = 2

y_true = np.array([4, 6, 8])

y_pred = X @ w + b

mse = np.mean((y_true - y_pred) ** 2)

print("Predictions:", y_pred)
print("MSE:", mse)
```

```python
#C7. Given a grayscale image as a 2D NumPy array with pixel values in [0, 255], normalize the values to [0, 1] using min-max normalization. 
import numpy as np

image = np.array([[0, 64],
                  [128, 255]])

normalized = (image - image.min()) / (image.max() - image.min())

print(normalized)
```

```python
#D1. Data Preprocessing Pipeline 
#You are given a raw 1D or 2D NumPy array containing NaN values.
#Describe, step by step, the NumPy operations you would apply to:
#(a) detect missing values,
#(b) handle them appropriately and justify your strategy, and
#(c) normalize or standardize the clean data before it is fed to a model.

#Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.



import numpy as np

data = np.array([
    [10.0, 20.0],
    [15.0, np.nan],
    [20.0, 30.0],
    [25.0, 40.0]
])

# Step 1: Detect missing values
missing = np.isnan(data)
print("Missing values:")
print(missing)

# Step 2: Replace NaN values with column means
column_means = np.nanmean(data, axis=0)
data_clean = np.where(missing, column_means, data)

print("Clean data:")
print(data_clean)

# Step 3: Standardize the clean data
mean = np.mean(data_clean, axis=0)
std = np.std(data_clean, axis=0)

standardized = (data_clean - mean) / std

print("Standardized data:")
print(standardized)
```
