```python
#1. Load the CSV file and inspect the dataset using head(), info(), and isnull().sum().
import pandas as pd

# Load CSV file
df = pd.read_csv("student_assessment_dirty.csv")

# Inspect the dataset
print(df.head())
print(df.info())
print(df.isnull().sum())
```

```python
#2.Standardize department names so that IT, it, and I.T. are represented consistently. Apply the same approach to Finance.

df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({
        "I.T.": "IT",
        "IT": "IT",
        "FINANCE": "Finance",
        "HR": "HR"
    })
)

# Check result
print(df["Department"].value_counts())

df["Department"] = df["Department"].str.strip()

df["Department"] = df["Department"].replace({
    "IT": "IT",
    "it": "IT",
    "I.T.": "IT",
    "Finance": "Finance",
    "finance": "Finance",
    "HR": "HR"
})

```

```python
#3.3.Identify and remove exact duplicate student assessment records.
# Find duplicate records
duplicates = df[df.duplicated(keep=False)]
print("Duplicate records:")
print(duplicates)

# Remove exact duplicates
df = df.drop_duplicates()

print("Number of records after removing duplicates:", len(df))
```

```python
#4.Convert the Marks column into a numeric format, handling "88 marks", "Absent", "NA", and blank entries appropriately.
# Convert marks to string first
df["Marks"] = df["Marks"].astype(str).str.strip()

# Convert "Absent", "NA", and blank values to missing
df["Marks"] = df["Marks"].replace({
    "Absent": pd.NA,
    "NA": pd.NA,
    "": pd.NA,
    "nan": pd.NA
})

# Remove text such as "marks"
df["Marks"] = df["Marks"].str.replace(
    r"\s*marks\s*", "", regex=True, case=False
)

# Convert to numeric
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print(df["Marks"])
```
